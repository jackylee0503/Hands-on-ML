```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedShuffleSplit

housing = pd.DataFrame({
    "median_income": [0.8, 1.0, 1.2, 1.4,   1.8, 2.2, 2.5, 2.9,
                      3.2, 3.6, 3.8, 4.1,   4.7, 5.0, 5.3, 5.6,
                      6.7, 7.8, 8.9, 10.1]  
})
housing["income_cat"] = pd.cut(
    housing["median_income"],
    bins=[0., 1.5, 3.0, 4.5, 6., np.inf],
    labels=[1, 2, 3, 4, 5],
)

splitter = StratifiedShuffleSplit(n_splits=10, test_size=0.25, random_state=42)  # 테스트 5개 = 그룹 수
for train_index, test_index in splitter.split(housing, housing["income_cat"]):
    strat_train_set_n = housing.iloc[train_index]
    strat_test_set_n = housing.iloc[test_index]

#print(strat_test_set_n["income_cat"].value_counts().sort_index())
#print(splitter.get_n_splits(housing, housing["income_cat"]))  # 10
print(housing["income_cat"])
print(strat_test_set_n["income_cat"])
```
